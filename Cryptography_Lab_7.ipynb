Columnar Transposition Cipher

```python
def columnar_encrypt(text, key):
    text = text.replace(" ", "").upper()
    key = key.upper()

    if not text.isalpha() or not text.isascii():
        raise ValueError("Text must contain English letters only.")

    if not key.isalpha() or not key.isascii():
        raise ValueError("Key must contain English letters only.")

    columns = len(key)
    rows = (len(text) + columns - 1) // columns

    grid = [text[i:i + columns] for i in range(0, len(text), columns)]

    order = sorted(range(columns), key=lambda i: (key[i], i))

    ciphertext = ""

    for col in order:
        for row in grid:
            if col < len(row):
                ciphertext += row[col]

    return ciphertext


def columnar_decrypt(ciphertext, key):
    ciphertext = ciphertext.replace(" ", "").upper()
    key = key.upper()

    if not ciphertext.isalpha() or not ciphertext.isascii():
        raise ValueError("Ciphertext must contain English letters only.")

    if not key.isalpha() or not key.isascii():
        raise ValueError("Key must contain English letters only.")

    columns = len(key)
    length = len(ciphertext)
    rows = (length + columns - 1) // columns
    remainder = length % columns

    order = sorted(range(columns), key=lambda i: (key[i], i))

    column_lengths = [
        rows if remainder == 0 or col < remainder else rows - 1
        for col in range(columns)
    ]

    grid = [""] * columns
    position = 0

    for col in order:
        size = column_lengths[col]
        grid[col] = ciphertext[position:position + size]
        position += size

    plaintext = ""

    for row in range(rows):
        for col in range(columns):
            if row < len(grid[col]):
                plaintext += grid[col][row]

    return plaintext


# Input validation
text = input("Enter plaintext: ").strip()

while not text or not text.isascii() or not text.replace(" ", "").isalpha():
    print("Error: Enter English letters and spaces only.")
    text = input("Enter plaintext: ").strip()

while True:
    key = input("Enter keyword: ").strip()

    if key.isascii() and key.isalpha():
        break

    print("Error: Keyword must contain English letters only.")

ciphertext = columnar_encrypt(text, key)

print("\nEncrypted text:", ciphertext)

print("Decrypted text:", columnar_decrypt(ciphertext, key))
```

Double Columnar Transposition Cipher

```python
def double_columnar_encrypt(text, key1, key2):
    first_encryption = columnar_encrypt(text, key1)
    second_encryption = columnar_encrypt(first_encryption, key2)

    return second_encryption


def double_columnar_decrypt(ciphertext, key1, key2):
    first_decryption = columnar_decrypt(ciphertext, key2)
    second_decryption = columnar_decrypt(first_decryption, key1)

    return second_decryption


# Input validation
text = input("Enter plaintext: ").strip()

while not text or not text.isascii() or not text.replace(" ", "").isalpha():
    print("Error: Enter English letters and spaces only.")
    text = input("Enter plaintext: ").strip()

while True:
    key1 = input("Enter first keyword: ").strip()

    if key1.isascii() and key1.isalpha():
        break

    print("Error: First keyword must contain English letters only.")

while True:
    key2 = input("Enter second keyword: ").strip()

    if key2.isascii() and key2.isalpha():
        break

    print("Error: Second keyword must contain English letters only.")

ciphertext = double_columnar_encrypt(text, key1, key2)

print("\nEncrypted text:", ciphertext)

plaintext = double_columnar_decrypt(ciphertext, key1, key2)

print("Decrypted text:", plaintext)
```
